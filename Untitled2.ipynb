```python
from ultralytics import YOLO

# 1. Download the lightweight YOLO model
model = YOLO('yolo11n.pt')

# 2. Start training on drone photos
results = model.train(
    data='VisDrone.yaml',
    epochs=50,
    imgsz=640,
    device=0,
    project='Edge_Quantization',
    name='baseline_fp32'
)
```

```python
from ultralytics import YOLO

# 1. Corrected path: pointing exactly to where the file lives inside runs/detect/
model = YOLO('runs/detect/Edge_Quantization/baseline_fp32/weights/best.pt')

# 2. Compress it to 8-bit integers (INT8)
# This will take a few minutes as it calibrates the 8-bit weights using the dataset
model.export(format="tflite", int8=True, data="VisDrone.yaml")
```

```python
from ultralytics import YOLO

# 1. Load the brand new, compressed 8-bit model
model_int8 = YOLO('runs/detect/Edge_Quantization/baseline_fp32/weights/best_int8.tflite')

# 2. Force it to take the exact same test the heavy model took
results_int8 = model_int8.val(data='VisDrone.yaml')
```
